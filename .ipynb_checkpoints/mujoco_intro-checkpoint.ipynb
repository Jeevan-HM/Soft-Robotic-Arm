# Building a Dynamic Model in MuJoCo

This notebook walks through the essential steps to create and simulate a **dynamic model** in MuJoCo from scratch using Python.

We'll cover:
1. The MJCF XML format (how you describe a model)
2. Bodies, joints, and geoms — the three building blocks
3. Loading the model and running a simulation
4. Reading state (position, velocity) and rendering frames
5. A slightly more complex example — a 2-link arm

No physics background required. We'll build intuition as we go.

## 1. The Three Building Blocks

Every MuJoCo model is described in **MJCF** (MuJoCo XML format). The three things you need to know first:

| Element | What it is | Analogy |
|---------|------------|---------|
| **Body** | A rigid chunk of matter with a position/orientation in space | A bone |
| **Geom** | A shape (box, sphere, capsule…) attached to a body for collision & rendering | Flesh on the bone |
| **Joint** | A degree of freedom connecting a child body to its parent | A hinge or socket |

Bodies are **nested**: a child body's position is relative to its parent. The root parent is always `worldbody`.

## 2. Example 1 — A Simple Pendulum

The classic: a rod that swings freely under gravity.

```
worldbody
  └── bob (body)
        joint: hinge about Y axis   ← 1 DOF
        geom:  capsule (the rod)
        geom:  sphere  (the ball)
```

We write this as an XML string and load it directly — no files needed.

````python
import mujoco
import numpy as np

# ── Define the model as an XML string ────────────────────────────────────────
PENDULUM_XML = """
<mujoco model="pendulum">

  <!-- Simulation settings -->
  <option timestep="0.005" gravity="0 0 -9.81"/>

  <worldbody>
    <!-- The pivot point is fixed at the world origin -->
    <!-- 'bob' hangs below it via a hinge joint -->
    <body name="bob" pos="0 0 0">

      <!-- A hinge joint: rotates freely about the Y axis -->
      <!-- damping adds a little air resistance so it eventually stops -->
      <joint name="swing" type="hinge" axis="0 1 0" damping="0.05"/>

      <!-- The rod: a capsule from the pivot (0,0,0) down to (0,0,-0.5) -->
      <geom name="rod" type="capsule"
            fromto="0 0 0  0 0 -0.5"
            size="0.01" mass="0.1"
            rgba="0.6 0.6 0.8 1"/>

      <!-- The ball at the end -->
      <geom name="ball" type="sphere"
            pos="0 0 -0.5" size="0.04"
            mass="0.5"
            rgba="0.9 0.3 0.2 1"/>

    </body>
  </worldbody>

</mujoco>
"""

# ── Load the model ────────────────────────────────────────────────────────────
model = mujoco.MjModel.from_xml_string(PENDULUM_XML)
data  = mujoco.MjData(model)

print(f"Degrees of freedom : {model.nq}")
print(f"Number of bodies   : {model.nbody}")
print(f"Number of geoms    : {model.ngeom}")
print(f"Timestep           : {model.opt.timestep} s")
````

### Key objects: `MjModel` vs `MjData`

| Object | Contains | Changes during sim? |
|--------|----------|---------------------|
| `MjModel` | The **blueprint** — geometry, masses, joint limits, … | No (static) |
| `MjData`  | The **state** — positions (`qpos`), velocities (`qvel`), forces, sensors | Yes (every step) |

Think of `MjModel` as the CAD drawing and `MjData` as the live sensor readout.

````python
# ── Set an initial condition and run one step ─────────────────────────────────

# qpos holds joint positions — our pendulum has 1 DOF (the 'swing' hinge)
# Let's start at 45° (pi/4 radians)
data.qpos[0] = np.pi / 4

# Always call mj_forward after manually setting qpos/qvel.
# It propagates positions through the kinematic chain and updates all derived
# quantities (body positions, Jacobians, etc.) without advancing time.
mujoco.mj_forward(model, data)

print("Before stepping:")
print(f"  qpos (angle)    = {data.qpos[0]:.4f} rad  ({np.degrees(data.qpos[0]):.1f} deg)")
print(f"  qvel (ang. vel) = {data.qvel[0]:.4f} rad/s")
print(f"  time            = {data.time:.4f} s")

# Advance physics by one timestep (0.005 s)
mujoco.mj_step(model, data)

print("\nAfter one step:")
print(f"  qpos (angle)    = {data.qpos[0]:.4f} rad  ({np.degrees(data.qpos[0]):.1f} deg)")
print(f"  qvel (ang. vel) = {data.qvel[0]:.4f} rad/s")
print(f"  time            = {data.time:.4f} s")
````

## 3. Running a Full Simulation

Stepping one tick at a time lets you record the state at each moment — this is how you build trajectories, plots, or video.

````python
import matplotlib.pyplot as plt

# Reset to a clean state, set initial angle to 90°
mujoco.mj_resetData(model, data)
data.qpos[0] = np.pi / 2   # 90 degrees — horizontal
mujoco.mj_forward(model, data)

# Simulate for 5 seconds and record angle + angular velocity
duration = 5.0
n_steps  = int(duration / model.opt.timestep)

t_log     = np.zeros(n_steps)
angle_log = np.zeros(n_steps)
vel_log   = np.zeros(n_steps)

for i in range(n_steps):
    t_log[i]     = data.time
    angle_log[i] = data.qpos[0]     # joint position [rad]
    vel_log[i]   = data.qvel[0]     # joint velocity [rad/s]
    mujoco.mj_step(model, data)     # advance physics

# Plot
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 5), sharex=True)

ax1.plot(t_log, np.degrees(angle_log), color="#4ECDC4", lw=1.5)
ax1.set_ylabel("Angle [deg]")
ax1.set_title("Simple Pendulum — free swing from 90 deg")
ax1.axhline(0, color="gray", lw=0.7, ls="--")
ax1.grid(alpha=0.3)

ax2.plot(t_log, np.degrees(vel_log), color="#FF6B6B", lw=1.5)
ax2.set_ylabel("Angular velocity [deg/s]")
ax2.set_xlabel("time [s]")
ax2.axhline(0, color="gray", lw=0.7, ls="--")
ax2.grid(alpha=0.3)

plt.tight_layout()
plt.show()
print("Damping causes the pendulum to gradually settle at 0 degrees.")
````

## 4. Rendering a Frame

MuJoCo renders **offscreen** by default — you get a NumPy array of pixels (no window needed).

`mujoco.Renderer` wraps an offscreen OpenGL context. You update the scene, then call `.render()` to get an `(H, W, 3)` uint8 RGB image.

````python
# Reset to a nice angle for visualisation
mujoco.mj_resetData(model, data)
data.qpos[0] = np.pi / 3    # 60 degrees
mujoco.mj_forward(model, data)

# Create an offscreen renderer (height x width in pixels)
renderer = mujoco.Renderer(model, height=400, width=400)

# Position the camera
cam = mujoco.MjvCamera()
cam.azimuth   = 90       # look from the side
cam.elevation = -10      # slightly downward
cam.distance  = 1.2      # 1.2 m away
cam.lookat[:] = [0, 0, -0.25]   # look at the mid-point of the rod

# Render!
renderer.update_scene(data, camera=cam)
frame = renderer.render()    # returns (400, 400, 3) uint8 array

plt.figure(figsize=(4, 4))
plt.imshow(frame)
plt.axis("off")
plt.title("Pendulum at 60 degrees")
plt.tight_layout()
plt.show()
print(f"Frame shape: {frame.shape}  dtype: {frame.dtype}")
````

## 5. Joint Types — Cheat Sheet

MuJoCo supports four joint types:

| Type | DOFs | Use for |
|------|------|---------|
| `hinge` | 1 rotation | Elbow, knee, pendulum |
| `slide` | 1 translation | Piston, linear actuator |
| `ball` | 3 rotations (quaternion) | Shoulder, hip |
| `free`  | 6 (3 pos + 3 rot) | Freely flying objects |

You can **stack** multiple joints on the same body to combine DOFs — for example, a `slide` + two `hinge`s gives a 3-DOF joint like in the soft arm (one extension + two bending axes per level).

## 6. Example 2 — A 2-Link Planar Arm

Now let's build something more interesting: a 2-DOF planar robot arm (like a shoulder + elbow).

```
worldbody
  └── link1  (upper arm, hinge at shoulder)
        └── link2  (forearm, hinge at elbow)
              site: tip  (end-effector reference point)
```

We'll also add a **site** — a massless reference point used to track the end-effector, and **sensors** to read it automatically.

````python
ARM_XML = """
<mujoco model="2link_arm">
  <option timestep="0.005" gravity="0 0 -9.81"/>

  <worldbody>

    <!-- Fixed shoulder pivot -->
    <body name="link1" pos="0 0 0">

      <!-- Shoulder hinge: rotates in the XZ plane -->
      <joint name="shoulder" type="hinge" axis="0 1 0"
             stiffness="0.2" damping="0.1"/>

      <!-- Upper arm: 0.3 m long, pointing down along -Z -->
      <geom name="upper_arm" type="capsule"
            fromto="0 0 0  0 0 -0.30"
            size="0.018" mass="0.3"
            rgba="0.3 0.55 0.9 1"/>

      <!-- Elbow: child body at the end of the upper arm -->
      <body name="link2" pos="0 0 -0.30">

        <!-- Elbow hinge: also rotates about Y -->
        <joint name="elbow" type="hinge" axis="0 1 0"
               stiffness="0.1" damping="0.08"/>

        <!-- Forearm: 0.25 m long -->
        <geom name="forearm" type="capsule"
              fromto="0 0 0  0 0 -0.25"
              size="0.014" mass="0.2"
              rgba="0.9 0.5 0.2 1"/>

        <!-- End-effector site: massless reference point at the tip -->
        <site name="tip" pos="0 0 -0.25" size="0.02" rgba="1 0 0 1"/>

      </body>
    </body>
  </worldbody>

  <!-- Sensors: read tip position and velocity automatically each step -->
  <sensor>
    <framepos    name="tip_pos" objtype="site" objname="tip"/>
    <framelinvel name="tip_vel" objtype="site" objname="tip"/>
  </sensor>

</mujoco>
"""

arm_model = mujoco.MjModel.from_xml_string(ARM_XML)
arm_data  = mujoco.MjData(arm_model)

print(f"DOFs        : {arm_model.nq}  (shoulder + elbow)")
print(f"Bodies      : {arm_model.nbody}")
print(f"Sensors     : {arm_model.nsensor}  (tip_pos, tip_vel)")
print(f"Sensor data : {arm_model.nsensordata} floats  (3 pos + 3 vel)")
````

````python
# ── Simulate the arm falling freely from a horizontal pose ───────────────────

mujoco.mj_resetData(arm_model, arm_data)

# Set shoulder to 90 degrees (horizontal), elbow straight (0 degrees)
arm_data.qpos[0] = np.pi / 2   # shoulder
arm_data.qpos[1] = 0.0          # elbow
mujoco.mj_forward(arm_model, arm_data)

duration = 3.0
n_steps  = int(duration / arm_model.opt.timestep)

t_log   = np.zeros(n_steps)
q_log   = np.zeros((n_steps, 2))   # joint angles
tip_log = np.zeros((n_steps, 3))   # tip XYZ position

for i in range(n_steps):
    t_log[i]   = arm_data.time
    q_log[i]   = arm_data.qpos[:2]
    # Sensor data: first 3 floats = tip position (from framepos sensor)
    tip_log[i] = arm_data.sensordata[0:3]
    mujoco.mj_step(arm_model, arm_data)

# Plot joint angles
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

ax = axes[0]
ax.plot(t_log, np.degrees(q_log[:, 0]), label="shoulder", color="#4ECDC4", lw=1.5)
ax.plot(t_log, np.degrees(q_log[:, 1]), label="elbow",    color="#FF6B6B", lw=1.5)
ax.set_title("Joint angles over time")
ax.set_xlabel("time [s]")
ax.set_ylabel("angle [deg]")
ax.legend()
ax.grid(alpha=0.3)
ax.axhline(0, color="gray", lw=0.7, ls="--")

ax = axes[1]
ax.plot(tip_log[:, 0], tip_log[:, 2], color="#C77DFF", lw=1.5)
ax.scatter(tip_log[0, 0],  tip_log[0, 2],  color="green", zorder=5, label="start", s=60)
ax.scatter(tip_log[-1, 0], tip_log[-1, 2], color="red",   zorder=5, label="end",   s=60)
ax.set_title("Tip trajectory (side view: X vs Z)")
ax.set_xlabel("X [m]")
ax.set_ylabel("Z [m]")
ax.set_aspect("equal")
ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
````

## 7. Reading State and Sensor Data

`data.sensordata` is a flat NumPy array. The layout matches the order you declared `<sensor>` elements in the XML.

You can also look up body positions directly from `data.xpos` — one row per body, indexed by body ID.

````python
# Reset to a clean pose and call mj_forward to populate all derived quantities
mujoco.mj_resetData(arm_model, arm_data)
arm_data.qpos[0] = np.pi / 4    # shoulder at 45 degrees
arm_data.qpos[1] = -np.pi / 4   # elbow bent back 45 degrees
mujoco.mj_forward(arm_model, arm_data)

# ── Sensor data ───────────────────────────────────────────────────────────────
tip_pos = arm_data.sensordata[0:3]   # framepos sensor output
tip_vel = arm_data.sensordata[3:6]   # framelinvel sensor output
print("Tip position (from sensor):", np.round(tip_pos, 4), "m")
print("Tip velocity (from sensor):", np.round(tip_vel, 4), "m/s")

# ── Body positions ────────────────────────────────────────────────────────────
# data.xpos shape: (nbody, 3). Row 0 is always the world body.
print("\nAll body positions (xpos):")
for i in range(arm_model.nbody):
    name = mujoco.mj_id2name(arm_model, mujoco.mjtObj.mjOBJ_BODY, i)
    print(f"  [{i}] {name:12s}  {np.round(arm_data.xpos[i], 4)}")

# ── Joint state ───────────────────────────────────────────────────────────────
print("\nJoint positions (qpos):", np.round(arm_data.qpos, 4), "rad")
print("Joint velocities (qvel):", np.round(arm_data.qvel, 4), "rad/s")
````

## 8. Render the 2-Link Arm

````python
arm_renderer = mujoco.Renderer(arm_model, height=400, width=500)

cam = mujoco.MjvCamera()
cam.azimuth   = 90
cam.elevation = -5
cam.distance  = 1.0
cam.lookat[:] = [0.1, 0, -0.3]   # look near the middle of the arm

# Render at the current pose (shoulder 45 deg, elbow -45 deg)
arm_renderer.update_scene(arm_data, camera=cam)
frame = arm_renderer.render()

plt.figure(figsize=(5, 4))
plt.imshow(frame)
plt.axis("off")
plt.title("2-link arm: shoulder=45 deg, elbow=-45 deg")
plt.tight_layout()
plt.show()
````

## 9. The Simulation Loop — a Clean Template

Here's the minimal pattern you'll use in every MuJoCo project:

```python
model = mujoco.MjModel.from_xml_string(XML)   # load once
data  = mujoco.MjData(model)                   # allocate state

# --- set initial conditions ---
data.qpos[:] = ...               # joint positions
data.qvel[:] = ...               # joint velocities  (optional)
mujoco.mj_forward(model, data)   # propagate kinematics

# --- simulation loop ---
for step in range(n_steps):
    # (optionally) read state
    state = data.qpos.copy()

    # (optionally) apply generalised forces directly
    data.qfrc_applied[:] = ...

    mujoco.mj_step(model, data)   # advance physics by one timestep
```

That's the whole loop. Everything else is book-keeping around it.

## Summary

| Task | How |
|------|-----|
| Define a model | XML string with `<body>`, `<geom>`, `<joint>` |
| Load it | `mujoco.MjModel.from_xml_string(xml)` |
| Allocate state | `mujoco.MjData(model)` |
| Set initial pose | `data.qpos[i] = value` then `mj_forward` |
| Step physics | `mujoco.mj_step(model, data)` |
| Read joint state | `data.qpos`, `data.qvel` |
| Read body positions | `data.xpos[body_id]` |
| Read sensor values | `data.sensordata` |
| Render | `Renderer` → `update_scene` → `render()` → NumPy RGB array |

The soft arm in this repo builds on exactly these primitives — it just has 15 DOFs (5 levels × 3 joints each) instead of 2, and writes pneumatic forces to `data.qfrc_applied` each step instead of letting the arm swing freely.
